```python
import pandas as pd

df = pd.read_csv('Ride_Booking_Data.csv')
```

```python
# Summary statistics using .describe()
df.describe(include='all')
```

```python
# Checking if missing data or null values are present in the dataset

df.isnull().sum()
```

```python
# 1. AvgVTAT

df['AvgVTAT'] = df['AvgVTAT'].fillna(df['AvgVTAT'].median())

# 2. Avg CTAT

df['Avg CTAT'] = df['Avg CTAT'].fillna(df['Avg CTAT'].median())

# 3. Reason for cancelling by Customer

df['Reason for cancelling by Customer'] = (
    df['Reason for cancelling by Customer']
    .fillna('Not Applicable')
)

# 4. Reason for cancelling by Driver

df['Reason for cancelling by Driver'] = (
    df['Reason for cancelling by Driver']
    .fillna('Not Applicable')
)

# 5. Incomplete Rides Reason

df['Incomplete Rides Reason'] = (
    df['Incomplete Rides Reason']
    .fillna('Not Applicable')
)

# 6. Booking Value

df['Booking Value'] = (
    df['Booking Value']
    .fillna(df['Booking Value'].median())
)

# 7. Ride Distance

df['Ride Distance'] = (
    df['Ride Distance']
    .fillna(df['Ride Distance'].median())
)

# 8. Driver Ratings

df['Driver Ratings'] = (
    df['Driver Ratings']
    .fillna(df['Driver Ratings'].mean())
)

# 9. Customer Rating

df['Customer Rating'] = (
    df['Customer Rating']
    .fillna(df['Customer Rating'].mean())
)
```

```python
df.isnull().sum()
```

```python
# snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
import numpy as np
# Payment methods
payment_methods = [
    "Cash",
    "UPI",
    "Credit Card",
    "Debit Card",
    "Wallet"
]

# Add new column
df["payment_method"] = np.random.choice(
    payment_methods,
    size=len(df),
    p=[0.25, 0.45, 0.12, 0.10, 0.08]
)
```
